### 1. 딕셔너리(Dictionary) 자료구조

- key - value 쌍 구조
- key : value를 찾기 위한 키워드, 중복 x, 리스트는 사용 불가
- value : key로 찾는 값, 중복 o
- {"키": 값, "키": 값, ...}

- 추가(CREATE): 없는 키 이름으로 값을 대입하면 추가됨
- 조회(READ): ['키 이름']
    - keys() : 모든 키를 담은 View, 읽기 전용
    - values() : 모든 값을 담은 View, 읽기 전용
    - imems() : (키, 값) 목록을 담은 View

```python
user = {
        "email": "test@test.org", 
        "password": 1234, 
        "name": "김철수"
    }
```

```python
user['mobile'] = "010-1000-1000"

user
```

```python
user["name"]
```

```python
user.keys()
```

```python
user.values()
```

```python
user.items()
```
