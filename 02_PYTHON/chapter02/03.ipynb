# 2장 3강 : 연산자

### 1. 항과 연산자

- 항: 연산에 사용되는 값, 2개 -> 2항, 3개 -> 3항, 1개 -> 단항
- 연산자: 연산에 사용되는 기호(+, =, -, ...)

### 2. 연산자의 종류

#### 대입 연산자(=)
x = 10
x = 10 + 20

연산의 우선 순위가 가장 낮다. 가장 마지막에 연산된다.


#### 산술 연산자
+, -
* : 곱하기
/ : 나누기
// : 나눈 몫
% : 나머지
** : 거듭제곱


- 연산 우선 순위: **, *, /, //, % > +, -
- 우선 순위 강제 적용: (..)

```python
num1 = 10
num2 = 3

num1 // num2, num1 % num2
```

```python
num1 = 3

num1 ** 3
```

#### 비교 연산자
>, <. >=, <=, 
== : 같다(동등성, 동일한 가치)
!= : 다르다

- 연산의 결과는 무조건 True, False
- 판별이 필요한 구문 (조건문, 반복문)

동등성 vs 동일성
- 동등성 : 동일한 가치를 비교(==)
- 동일성 : 동일한 대상인지 비교(is - 주소 비교)

```python
10 > 5
```

```python
num1 = 1000
num2 = 1000

id(num1), id(num2)
```

```python
num1 == num2 # 동등성 비교
```

```python
num1 is num2 # 동일성 비교(주소 비교)
```

```python
num3 = num1

num3 is num1
```

```python
num1 = 20

result1 = num1 > 10
result2 = num1 <100

result1, result2
```

#### 논리 연산자 (and, or, not)

and
- 각 항이 모두 True -> True, 하나라도 False -> False

or
- 각 항의 하나라도 True -> True, 모두 False -> False

not
- True 조건 -> False 조건으로 반전, False 조건 -> True 조건으로 반전

비교 연산자 보다 우선순위가 낮다

```python
# and
num1 = 20

result1 = num1 > 10
result2 = num1 <100

result1 and result2
```

```python
not (result1 and result2)
```

```python
fruit1 = "사과"
fruit2 = "배"

result1 = fruit1 == "사과"
result2 = fruit2 == "오렌지"

result1, result2
```

```python
result1 or result2
```

```python
fruit1 == "사과" or fruit2 == "오렌지"
```

#### 복합 대입 연산자
대입 연산자(=) + 다른 연산자(주로 산술 연산자 +, -, *, /* //* %, **) 결합


```python
num = 1
num = num + 1
num = num + 1
num = num + 1

num
```

```python
num = 1
num += 1  # 단항 연산, num = num + 1
num += 1
num += 1

num
```

```python
num *=2

num
```

#### 부호 반전 연산자(-)

음수 -> 양수
양수 -> 음수

```python
num = 100

num2 = -num
num2
```

```python
num3 = -num2 # 음수 -> 양수
num3
```

### 3. 단락 회로 평가(Short-Circuit Evaluation)

```python
result = True or 10**2 # 첫 항의 결과가 True -> or 연산의 결과가 이미 결정! -> 다음 연산 진행 x
result
```

```python
result = False or 10**2 # 첫 항의 결과가 False -> 결과가 아직 결정 x -> 다음 항 체크(연산)
result
```

```python
result = False and 10**2
result
```

```python
result = True and 10**2
result
```

```python
# or 연산을 이용한 기본값(없을 때 대체되는 값)
message = ""

result = message or "기본 메시지"
result
```

```python
# and 연산을 이용한 간단한 조건 처리
height = 165

result = height >= 160 and "크다"
result
```

### 4. 연산자 우선순위

(..) > 산술 연산자 > 비교 연산자 > 논리 연산자 > 대입 연산자(=)
