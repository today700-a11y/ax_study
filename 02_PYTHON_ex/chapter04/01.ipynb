```python
weight = 10
if weight > 23:
    print("대형 수화물 대상입니다. 추가 요금이 부과됩니다.")
elif weight > 15:
    print("일반 위탁 수하물 대상입니다.")
else:
    print("무료 수하물 대상입니다.")
```
